# Overall steps to use our repo

1. Make a token on your personal git which is required for push
2. Use the token to clone the repo
3. Create your personal branch on the git
4. Push something to your branch





# Generate token

Steps to create a PAT:
Go to your GitHub account.
In the top-right corner, click your profile picture, and then click Settings.
In the left sidebar, click Developer settings.
Under Developer settings, click Personal access tokens.
Click Generate new token.
Give it a name (e.g., colab-token), select the scopes (e.g., repo), and click Generate token.
Copy the token (you won't be able to see it again once you leave this page
